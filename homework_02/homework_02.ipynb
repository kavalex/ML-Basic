## Задание 1: Получение однозначного числа

Пользователь вводит целое число, программа складывает все цифры числа, с полученным числом — то же самое, и так до тех пор, пока не получится однозначное число.

<br>

**Примеры:**
* `545   -> 5`
* `12345 -> 6`

````python
try:
    n = abs(int(input()))
    while n >= 10:
        n = sum(int(digit) for digit in str(n))
        print(n)
except ValueError:
    print("Неверный ввод")

````

## Задание 2: Кинотеатр

Дан список списков, каждый вложенный список состоит из 1 и 0, количество вложенных списков — количество рядов. Пользователь вводит, сколько билетов ему требуется. Программа должна найти ряд, где можно приобрести нужное количество билетов (места должны быть рядом). Если таких рядов несколько, то ближайший к экрану (ближайшим считается нулевой ряд). Если таких мест нет, то вывести False.

<br>

**Примеры:**
* `[[0,1,1,0], [1,0,0,0], [0,1,0,0]], 2 -> 1`
* `[[0,1,1,0], [1,0,1,0], [1,1,0,1]], 2 -> False`

````python
import ast

def max_consecutive_free(row):
    max_count = 0
    current_count = 0
    for seat in row:
        if seat == 0:
            current_count += 1
            if current_count > max_count:
                max_count = current_count
        else:
            current_count = 0
    return max_count

def find_row(cinema, tickets):
    # Валидация
    if not isinstance(cinema, (list, tuple)):
        raise ValueError("Кинозал должен быть списком или кортежем рядов")
    for row in cinema:
        if not isinstance(row, (list, tuple)):
            raise ValueError("Каждый ряд должен быть списком или кортежем мест")
        for cell in row:
            if cell not in (0, 1):
                raise ValueError("Ячейки должны быть 0 или 1")
    if tickets <= 0:
        return False

    # Поиск первого подходящего ряда
    for i, row in enumerate(cinema):
        if max_consecutive_free(row) >= tickets:
            return i
    return False

# Ввод данных
raw = input("Введите информацию о наличии мест по каждому ряду в кинотеатре в виде списка списков (например, [[0,1],[1,0]]): ").strip()

try:
    cinema = ast.literal_eval(raw)
except Exception as e:
    print("Неверный формат:", e)
else:
    try:
        tickets = int(input("Введите количество билетов: "))
    except ValueError:
        print("Неверный ввод количества билетов")
    else:
        try:
            result = find_row(cinema, tickets)
            print(result)
        except ValueError as e:
            print("Ошибка:", e)
````

## Задание 3: Алгоритм RLE

Необходимо написать упрощенную версию алгоритма RLE. 
Алгоритм RLE объединяет подряд идущие символы в коэффициент и символ.

<br>

**Примеры:**
* `asssdddsssddd -> 1a3s3d3s3d`
* `aaabbbbccccc  -> 3a4b5c`
* `abcba         -> 1a1b1c1b1a`

````python
s = input().strip()
if not s:
    print()
    exit()

result = []
current_char = s[0]
count = 1

for char in s[1:]:
    if char == current_char:
        count += 1
    else:
        result.append(f"{count}{current_char}")
        current_char = char
        count = 1
result.append(f"{count}{current_char}")  # Добавляем последний символ
print("".join(result))
````

## Задание 4: Шифр Цезаря

Реализуйте программу для шифрования текста с помощью шифра Цезаря.  
Шифр Цезаря — это метод шифрования, при котором каждая буква в тексте заменяется буквой, стоящей на фиксированное число позиций дальше в алфавите.

### Требования
1. Программа должна принимать на вход:
   - Строку текста для шифрования
   - Ключ шифра (число от 0 до 25)

2. Правила шифрования:
   - Шифруются только латинские буквы (a-z, A-Z)
   - Сохраняется регистр букв (заглавные остаются заглавными)
   - Пробелы, цифры и знаки препинания остаются без изменений
   - Сдвиг циклический (после 'z' идет 'a', после 'Z' идет 'A')

### Примеры

* `"Hello World", 3      -> "Khoor Zruog"`
* `"Python is great!", 5 -> "Udymts nx lwjfy!"`
* `"XYZ abc", 3          -> "ABC def"`
* `"Test 123", 1         -> "Uftu 123"`

### Алгоритм
* Для каждого символа в строке:
   - Если это латинская буква, найти её позицию в алфавите
   - Добавить к позиции значение ключа
   - Если результат больше 25, использовать остаток от деления на 26
   - Преобразовать обратно в букву, сохранив регистр
   - Если это не буква, оставить символ без изменений


````python
text = input().strip()
key = int(input())

def caesar(char, shift):
    if 'a' <= char <= 'z':
        return chr(((ord(char) - ord('a') + shift) % 26) + ord('a'))
    elif 'A' <= char <= 'Z':
        return chr(((ord(char) - ord('A') + shift) % 26) + ord('A'))
    return char

encrypted = "".join(caesar(char, key) for char in text)
print(encrypted)
````

## Задание 5: Табель успеваемости

Пользователь в бесконечном цикле (пока не будет введена пустая строка) вводит строки вида: 'название предмета' 'фамилия ученика' 'оценка'. После окончания ввода программа выводит в консоль название предмета, далее список учеников и все их оценки в виде таблицы.

<br>

**Примеры:**

<br>

*Ввод:*
```python
Математика Иванов  5
Математика Иванов  4
Литература Иванов  3
Математика Петров  5
Литература Сидоров 3
Литература Петров  5
Литература Иванов  4
Математика Сидоров 3
Математика Петров  5
```

<br>

*Результат:*
```python
Математика # вводим название предмета

# выводим список учеников и их оценки
Иванов  5 4
Петров  5 5
Сидоров 3
```

```python
Литература # вводим название предмета

# выводим список учеников и их оценки
Иванов  3 4
Сидоров 3
Петров  5
```

````python
from collections import defaultdict

# Словарь: { предмет -> { ученик -> [оценки] } }
subjects = defaultdict(lambda: defaultdict(list))

print("Введите строки в формате: 'предмет фамилия оценка'")
print("Для завершения введите пустую строку.")

while True:
    line = input().strip()
    if not line:
        break
    parts = line.rsplit(maxsplit=2)
    if len(parts) != 3:
        print("Неверный формат. Ожидается: 'предмет фамилия оценка'")
        continue
    subject, student, grade_s = parts
    try:
        grade = int(grade_s)
    except ValueError:
        print("Оценка должна быть числом.")
        continue
    if not (1 <= grade <= 5):
        print("Оценка должна быть в диапазоне 1..5.")
        continue
    subjects[subject][student].append(grade)

# Выводим результаты
for subject, students in subjects.items():
    print(subject)
    for student, grades in students.items():
        print(f"{student} {' '.join(map(str, grades))}")
    print()  # пустая строка между предметами
````
