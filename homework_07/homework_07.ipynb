### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
rng = np.random.default_rng(0)
a = rng.random((4,3,2))
print("a:\n", a)
print("shape:", a.shape)
min_val = a.min()
max_val = a.max()
print("min:", min_val)
print("max:", max_val)
```

```python
a = np.random.random((6, 4))
b = np.random.random((4, 3))
print(a)
print(b)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
a = rng.random((7, 7))
print(a)
print("shape:", a.shape)
mean = a.mean()
print("mean:", mean)
std = a.std()
print("std:", std)
a_norm = (a - mean) / std
print(a_norm)
print("mean(after):", a_norm.mean())
print("std(after):", a_norm.std())
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
mean_tb = tips['total_bill'].mean()
median_tb = tips['total_bill'].median()
plt.figure(figsize=(8,5))
sns.histplot(tips['total_bill'], bins=20, kde=True, color='C0')
plt.axvline(mean_tb, color='C1', linestyle='--', label=f"mean = {mean_tb:.2f}")
plt.axvline(median_tb, color='C2', linestyle=':', label=f"median = {median_tb:.2f}")
plt.xlabel('total_bill')
plt.ylabel('count')
plt.title('Histogram of total_bill')
plt.legend()
plt.tight_layout()
plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
from scipy import stats
r, p = stats.pearsonr(tips['total_bill'], tips['tip'])
print(f"Pearson r = {r:.3f}, p-value = {p:.3e}")
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex',        # раскрасить по полу
    style='smoker',   # разные маркеры для курильщиков/некурящих
    s=60,             # размер точек
    alpha=0.7,        # задать прозрачность точек, чтобы перекрытия были видны
    palette='Set1'
)
sns.regplot(data=tips, x='total_bill', y='tip',
            scatter=False,    # точки не рисовать, они уже есть
            color='gray',
            line_kws={'linestyle':'--', 'linewidth':1}
)
plt.xlabel('total_bill')
plt.ylabel('tip')
plt.title('Scatterplot: total_bill vs tip\n' + f'Pearson r = {r:.3f}')
plt.legend(title='sex / smoker', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Постройте pairplot

```python
sns.pairplot(tips)
plt.show()
num_cols = tips.select_dtypes(include='number').columns.tolist()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
print(tips['total_bill'].min())
print(tips['day'].unique())
print(tips['day'].value_counts().reindex(tips['day'].cat.categories))
# Boxplot
plt.figure(figsize=(8,5))
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Boxplot: total_bill by day')
plt.xlabel('day')
plt.ylabel('total_bill')
plt.tight_layout()
plt.show()
# Violinplot
plt.figure(figsize=(8,5))
sns.violinplot(data=tips, x='day', y='total_bill', cut=0, inner='quartile')
plt.title('Violinplot: total_bill by day')
plt.tight_layout()
plt.show()
# Stripplot - случайный разброс точек по горизонтали (jitter)
plt.figure(figsize=(8,5))
sns.boxplot(data=tips, x='day', y='total_bill', showfliers=False)
sns.stripplot(data=tips, x='day', y='total_bill', alpha=0.5, jitter=0.3)
plt.title('Boxplot + points')
plt.tight_layout()
plt.show()
# Swarmplot - расчетное смещение точек по горизонтали (beeswarm), для минимизации разброса
plt.figure(figsize=(8,5))
sns.boxplot(data=tips, x='day', y='total_bill', showfliers=False)
sns.swarmplot(data=tips, x='day', y='total_bill')
plt.title('Swarmplot: total_bill by day')
plt.tight_layout()
plt.show()
# Barplot
plt.figure(figsize=(8,5))
sns.barplot(data=tips, x='day', y='total_bill', errorbar=('ci', 95))
plt.title('Mean total_bill by day')
plt.tight_layout()
plt.show()
# Pointplot
plt.figure(figsize=(8,5))
sns.pointplot(data=tips, x='day', y='total_bill', capsize=.1)
plt.title('Pointplot: mean total_bill by day')
plt.tight_layout()
plt.show()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
# задаём одинаковые бины для корректного сравнения
bins = np.linspace(tips['tip'].min(), tips['tip'].max(), 16)

plt.figure(figsize=(8,5))
sns.histplot(
    data=tips,
    x='tip',
    hue='time',
    bins=bins,
    multiple='dodge',   # ставит бары рядом
    shrink=0.8,         # уменьшает ширину баров, чтобы между ними было пространство
    edgecolor='k',      # цвет границы - черный
    alpha=0.8
)
plt.xticks(bins)        # метки для каждой границы бина
plt.xlabel('tip')
plt.ylabel('count')
plt.title('Tip distribution: Lunch vs Dinner (side-by-side bars)')
plt.tight_layout()
plt.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
sns.set(style='whitegrid')
g = sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='sex',                # два графика: Male / Female
    hue='smoker',             # цвет точки по признаку smoker
    palette={'Yes':'C1','No':'C0'},
    kind='scatter',
    height=4,                 # высота каждой панели в дюймах, 4 inches
    aspect=1,                 # соотношение ширины к высоте
    facet_kws={'sharex': True, 'sharey': True} # одинаковые значения по x и по y
)

g.set_axis_labels('total_bill', 'tip')
g.set_titles(col_template='{col_name}')
g.fig.suptitle('Total bill vs Tip by Sex (colored by smoker)', y=1.02)
plt.show()
```

#### 4. Зависимость общего чека от размера компании, поле size
Есть одна пара, которая сделала заказ на более чем 40 долларов. Такой дорогой заказ делали немногие, даже большие компании. Например у компаний из 5 и 6 человек было всего по одному такому дорогому заказу.

```python
high = tips[tips['total_bill'] > 40].copy()
counts = high['size'].value_counts().sort_index()   # сортируем по size
colors = sns.color_palette('Blues_d', n_colors=len(counts))
plt.figure(figsize=(8,5))
plt.bar([str(x) for x in counts.index], counts.values, color=colors, edgecolor='k')
plt.xlabel('size')
plt.ylabel('count')
plt.title('Count of orders with total_bill > 40 by size')
ymax = int(counts.values.max())
plt.yticks(np.arange(0, ymax + 1, 1))   # метки 0,1,2,...,ymax
plt.tight_layout()
plt.show()
```

Что это, дорогое вино или парочка очень голодных людей?
Вечер воскресенья, платит джентельмен, скорее-всего это романтический ужин.

```python
subset = tips[tips['size'] == 2]
idx = subset['total_bill'].idxmax()
print(tips.loc[idx])
```

Есть компания из 6 человек, которая сделала рекордно низкий чек, платила дама.
Похоже подруги за обедом посидели за чашечкой кофе в будний день и обсудили последние новости.

```python
subset = tips[tips['size'] == 6]
idx = subset['total_bill'].idxmin()
print(tips.loc[idx])
```
