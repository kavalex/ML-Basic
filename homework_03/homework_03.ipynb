## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
s = str(input())
if s == "":
    print()

out = []
capitalize_next = False  # выставляется при встрече '_'
first_letter_handled = False  # отметка, что первая буква-символ уже обработана

length = len(s)
i = 0
while i < length:
    ch = s[i]

    if ch == '_':
        capitalize_next = True
        i += 1
        continue

    next_ch = s[i + 1] if i + 1 < length else None
    next_is_lower = next_ch.islower() if (next_ch is not None and next_ch.isalpha()) else False

    if ch.isupper():
        # Решение о вставке '_' зависит от следующего символа:
        # вставляем '_' перед строчной версией только если next_is_lower == True
        need_underscore = False
        if out and out[-1] != '_':
            # вставлять можно только если следующая буква в исходе строчная
            if next_is_lower:
                need_underscore = True

        if need_underscore:
            out.append('_')

        # Первая буква в выходе: применяем инверсию регистра (заглавная -> строчная)
        if not first_letter_handled and (ch.isalpha()):
            out.append(ch.lower())
            first_letter_handled = True
            capitalize_next = False
        else:
            out.append(ch.lower())
            capitalize_next = False

        i += 1
        continue

    if ch.islower():
        # Первая буква: строчная -> заглавная (инверсия первой буквы)
        if not first_letter_handled:
            out.append(ch.upper())
            first_letter_handled = True
            capitalize_next = False
            i += 1
            continue

        # Флаг от '_' делает текущую букву заглавной
        if capitalize_next:
            out.append(ch.upper())
            capitalize_next = False
        else:
            out.append(ch)
        i += 1
        continue

    if ch.isdigit():
        out.append(ch)
        capitalize_next = False
        i += 1
        continue

    # прочие символы — игнорируем и сбрасываем флаг
    capitalize_next = False
    i += 1

print(''.join(out))
```

## Задание 2: Проверка валидности даты

Написать функцию проверяющую валидность введенной даты.

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
from datetime import datetime

s = str(input())
try:
    datetime.strptime(s, "%d.%m.%Y")
    print(True)
except (ValueError, TypeError):
    print(False)
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
import math

n = int(input())
if n <= 1:
    print(False)
elif n <= 3:
    print(True)   # 2 и 3 — простые
elif n % 2 == 0 or n % 3 == 0:
    print(False)
else:
    # Проверяем делители вида 6k-1 и 6k+1
    limit = int(math.isqrt(n))
    i = 5
    while i <= limit:
        if n % i == 0 or n % (i + 2) == 0:
            print(False)
        i += 6
    print(True)
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
def read_nonempty_or_break(prompt: str) -> str:
    """Считывает строку; если пустая — возвращает '' для завершения ввода."""
    s = input(prompt).strip()
    return s

def valid_name_part(s: str) -> bool:
    """Проверка: только буквы (русские/латинские) и не пусто."""
    return s.isalpha() and len(s) > 0

def normalize_name_part(s: str) -> str:
    """Приводим к форме: первая заглавная, остальные строчные."""
    return s.capitalize()

def format_id(num: int) -> str:
    """Возвращает строку ID в виде 8 знаков с ведущими нулями."""
    return str(num).zfill(8)

def print_users_table(users: dict):
    """
    users: dict где ключ — id_str (8 символов), значение — (name, surname, age)
    Печатает таблицу столбцами: ID | Name | Surname | Age
    """
    if not users:
        print("Пользователи отсутствуют.")
        return

    # Подготовим строки и вычислим ширины колонок
    rows = []
    for id_str, (name, surname, age) in users.items():
        rows.append((id_str, name, surname, str(age)))

    # Сортируем по ID для предсказуемого вывода
    rows.sort(key=lambda r: r[0])

    headers = ("ID", "Name", "Surname", "Age")
    col_widths = [len(h) for h in headers]
    for row in rows:
        for i, cell in enumerate(row):
            col_widths[i] = max(col_widths[i], len(cell))

    # Печать заголовка
    sep = " | "
    header_line = sep.join(h.ljust(col_widths[i]) for i, h in enumerate(headers))
    divider = "-+-".join("-" * col_widths[i] for i in range(len(headers)))
    print(header_line)
    print(divider)
    # Печать строк
    for row in rows:
        print(sep.join(row[i].ljust(col_widths[i]) for i in range(len(row))))

users = {}  # ключ: id_str (8 символов), значение: (Name, Surname, age)

print("Ввод пользователей. Для завершения введите пустое поле (Enter) в любом запросе.")
while True:
    name = read_nonempty_or_break("Имя: ")
    if name == "":
        break
    surname = read_nonempty_or_break("Фамилия: ")
    if surname == "":
        break
    age_s = read_nonempty_or_break("Возраст: ")
    if age_s == "":
        break
    id_s = read_nonempty_or_break("ID (целое число): ")
    if id_s == "":
        break

    # Валидация имени/фамилии
    if not valid_name_part(name):
        print("Ошибка: имя должно содержать только буквы. Попробуйте снова.")
        continue
    if not valid_name_part(surname):
        print("Ошибка: фамилия должна содержать только буквы. Попробуйте снова.")
        continue

    # Нормализация: первая буква заглавная
    name_norm = normalize_name_part(name)
    surname_norm = normalize_name_part(surname)

    # Валидация возраста
    try:
        age = int(age_s)
    except ValueError:
        print("Ошибка: возраст должен быть целым числом. Попробуйте снова.")
        continue
    if not (18 <= age <= 60):
        print("Ошибка: возраст должен быть в диапазоне 18..60. Попробуйте снова.")
        continue

    # Валидация ID
    try:
        id_num = int(id_s)
        if id_num < 0:
            raise ValueError
    except ValueError:
        print("Ошибка: ID должен быть неотрицательным целым числом. Попробуйте снова.")
        continue
    id_str = format_id(id_num)
    if id_str in users:
        print(f"Ошибка: ID {id_str} уже существует. Введите уникальный ID. Попробуйте снова.")
        continue

    # Всё прошло — сохраняем
    users[id_str] = (name_norm, surname_norm, age)
    print(f"Пользователь {name_norm} {surname_norm} (возраст {age}) с ID {id_str} добавлен.\n")

# По завершении ввода выводим таблицу
print("\nСписок пользователей:")
print_users_table(users)
```
